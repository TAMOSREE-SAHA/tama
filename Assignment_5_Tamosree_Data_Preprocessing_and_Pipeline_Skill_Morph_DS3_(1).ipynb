**# 2.** **Separate X (features) and y (target = 'airconditioning')**

```python
X = df.drop('airconditioning', axis=1)  # Everything except airconditioning
y = df['airconditioning']                # Only airconditioning

print("X has all housing information (12 columns)")
print("y has housing airconditioning (1 column)")

```

**# 5. Train-test split (80% train, 20% test) — remember stratify=y**

```python


# Split: 80% for training, 20% for testing
# stratify=y keeps the same Outcome (yes/no) ratio in both train and test sets

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,    # 20% for test
    stratify=y,       # keep the same Outcome ratio in train and test
    random_state=30 # Same result every time
)

print(f"\nWe will train with: {len(X_train)} housings")
print(f"We will test with: {len(X_test)} housings")
```

**# 3.** **Label Encode: mainroad, guestroom, basement, hotwaterheating, prefarea, airconditioning**


```python
from sklearn.preprocessing import LabelEncoder

# For yes/no  (2 categories only)
label_encoder = LabelEncoder()

# Convert 'mainroad' column

X_train['mainroad'] = label_encoder.fit_transform(X_train['mainroad'])
X_test['mainroad'] = label_encoder.transform(X_test['mainroad'])
print("\nAfter Label Encoding 'mainroad':")
print(X_train['mainroad'])

# Convert 'guestroom' column

X_train['guestroom'] = label_encoder.fit_transform(X_train['guestroom'])
X_test['guestroom'] = label_encoder.transform(X_test['guestroom'])
print("\nAfter Label Encoding 'guestroom':")
print(X_train['guestroom'])

# Convert 'basement' column

X_train['basement'] = label_encoder.fit_transform(X_train['basement'])
X_test['basement'] = label_encoder.transform(X_test['basement'])
print("\nAfter Label Encoding 'basement':")
print(X_train['basement'])

```

```python
# Convert 'hotwaterheating' column

X_train['hotwaterheating'] = label_encoder.fit_transform(X_train['hotwaterheating'])
X_test['hotwaterheating'] = label_encoder.transform(X_test['hotwaterheating'])
print("\nAfter Label Encoding 'hotwaterheating':")
print(X_train['hotwaterheating'])

# Convert 'prefarea' column

X_train['prefarea'] = label_encoder.fit_transform(X_train['prefarea'])
X_test['prefarea'] = label_encoder.transform(X_test['prefarea'])
print("\nAfter Label Encoding 'prefarea':")
print(X_train['prefarea'])


# no = 0, yes = 1
```

```python
# Convert 'airconditioning' column

y_train = label_encoder.fit_transform(y_train)
y_test = label_encoder.transform(y_test)
print("\nAfter Label Encoding 'prefarea':")
print(y_train)
```

**# 4. One-Hot Encode: furnishingstatus**

```python

X_train=pd.get_dummies(X_train, columns=['furnishingstatus'], drop_first=True, dtype=int)
X_test=pd.get_dummies(X_test, columns=['furnishingstatus'], drop_first=True, dtype=int)

print(X_train)

```

**# 6.Scale the numeric features with StandardScaler**




```python
from sklearn.preprocessing import StandardScaler
# Create scaler
scaler = StandardScaler()

# Learn from training data and scale it
X_train_scaled = scaler.fit_transform(X_train)


# Scale test data (just transform, don't fit)
X_test_scaled = scaler.transform(X_test)
print("\nAfter scaling:")
print("All features now have similar range!")
print(f"Example - First patient's age before scaling: {X_train.values[0][0]}")
print(f"Example - First patient's age after scaling: {X_train_scaled[0][0]:.2f}")
```

**# 7.** **Run: train a LogisticRegression model, apply cross validation( 5 fold, 10 fold) and print the accuracy**

```python
# train a LogisticRegression model

from sklearn.linear_model import LogisticRegression

# Create model
model = LogisticRegression(max_iter=1000)

# Train the model
model.fit(X_train_scaled, y_train)
print("✅ Model trained!")

# Check accuracy on training data
train_score = model.score(X_train_scaled, y_train)
print(f"Training accuracy: {train_score:.1%}")

# Check accuracy on test data
test_score = model.score(X_test_scaled, y_test)
print(f"Test accuracy: {test_score:.1%}")
```

```python
from sklearn.model_selection import cross_val_score

# Do 5-fold cross-validation
scores = cross_val_score(
    model,           # Our model
    X_train_scaled,  # Training data
    y_train,         # Training labels
    cv=5            # 5 mini-tests
)

print("5 mini-test scores:")
for i in range(5):
  print(f"  Test {i+1}: {scores[i]:.1%}")

print(f"\nAverage: {scores.mean():.1%}")
print(f"This means our model is {scores.mean():.1%} accurate!")
```

```python
# Do 10-fold cross-validation

scores = cross_val_score(
    model,           # Our model
    X_train_scaled,  # Training data
    y_train,         # Training labels
    cv=10            # 10 mini-tests
)

print("5 mini-test scores:")
for i in range(10):
  print(f"  Test {i+1}: {scores[i]:.1%}")

print(f"\nAverage: {scores.mean():.1%}")
print(f"This means our model is {scores.mean():.1%} accurate!")
```
