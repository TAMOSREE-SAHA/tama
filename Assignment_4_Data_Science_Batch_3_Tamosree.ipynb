```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Sensible global defaults so every plot looks consistent
sns.set_theme(style='whitegrid', context='notebook')
plt.rcParams['figure.figsize'] = (7, 4)
plt.rcParams['figure.dpi']     = 100

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
students = pd.read_csv('/content/drive/MyDrive/Dataset/StudentsPerformance.csv')
gym      = pd.read_csv('/content/drive/MyDrive/Dataset/gym_membership.csv')


gym.head()

```

```python
# Q1. Display the first 5 rows of the dataset, then print its shape

print(students.shape)
students.head()
```

```python
# Q2. Print the average (mean) math score, reading score and writing score

avg_score = students[['math score', 'reading score', 'writing score']].mean()

avg_score = avg_score.reset_index()
avg_score.columns = ['    Subject', '      Average Score']

print(avg_score)
```

```python
# Q3. Display all the students who scored more than 90 in reading, and print how many such students there are.

high_reading = students[students['reading score'] > 90]

print(high_reading)
print("Number of students:", len(high_reading))

```

```python
# Q4. Sort the dataset by math score in descending order and display the top 5 rows.

sorted_students = students.sort_values('math score', ascending=False)

print(sorted_students.head())
```

```python
# Q5. Bar chart of membership types — gym['abonoment_type'].value_counts()

gym_membership= gym['abonoment_type'].value_counts()

fig, ax = plt.subplots()
ax.bar(gym_membership.index, gym_membership.values, color=['red', 'green'])
ax.set(title='Bar chart of membership types', ylabel='Count')
plt.show()
```

```python
gym.info()
```

```python
# Q6. Histogram of avg_time_in_gym

fig, ax = plt.subplots()
ax.hist(gym['avg_time_in_gym'], bins=9, color='green', edgecolor='blue')
ax.set(title='Average Time Passed in Gym',
       xlabel='Average Time', ylabel='Member')
plt.show()
```

```python
# Q7. Scatter plot of Age vs visit_per_week

fig, ax = plt.subplots()
ax.scatter(gym['Age'], gym['visit_per_week'],
           alpha=.9, s=15, color='green')
ax.set(title='Age VS Visit per Week', xlabel='Age', ylabel='Visit per Week')

ax.set_yticks([1, 2, 3, 4, 5])
plt.show()

```

```python
# Q8. Box plot — gender vs Age

fig, axes = plt.subplots()

sns.boxplot(data=gym, x='gender', y='Age', ax=axes)
axes.set_title('Age by Gender')

plt.tight_layout()
plt.show()
```

```python
# Q9. Count plot — days_per_week


fig, axes = plt.subplots(1, 2, figsize=(11, 4))

sns.countplot(data=gym, x='gender', ax=axes[0])
axes[0].set_title('Gender Vs Day per Week')

sns.countplot(data=gym, x='abonoment_type', ax=axes[1])
axes[1].set_title('Abonnement Vs Day per Week')

plt.tight_layout()
plt.show()
```

```python
# Q10. Bar plot — mean visit_per_week per abonoment_type

fig, ax = plt.subplots(figsize=(9, 4))
order = gym.groupby('abonoment_type')['visit_per_week'].mean().sort_values().index
sns.barplot(data=gym, x='abonoment_type',
            y='visit_per_week', order=order, ax=ax)
ax.set_title('Mean Visit per Week per Abonoment Type')
ax.tick_params(axis='x', rotation=10)
plt.tight_layout()
plt.show()
```
