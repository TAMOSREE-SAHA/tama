```python
## H8: Create a program that converts temperatures between Fahrenheit and Celsius based on user input.

print("For Fahrenheit to Celsius, press 1 \nfor Celsius to Fehrenheit, press 2 ")

n= int(input("Here put your choice :"))

if n == 1:
  f = float(input("Enter Fahrenheit value: "))
  c = (f - 32)*(5/9)
  print("Celsius Temperature : ",c)

elif n == 2:
  c = int(input("Enter your Celsius value: "))
  f = (c * 1.8) + 32
  print("Fahrenheit Temperature : ",f)

```

```python
# H10: Write a program that prompts the user to input a number N and prints all even numbers between 1 and N using a for loop.

N=int(input("Enter the integer N: "))
x=0

for x in range (N+1):
  if x % 2 == 0:
    print(x )
  x=x+1
```

```python
# H11: Write a program that prompts the user to input a number N and calculates the sum of all numbers from 1 to N that are divisible by 3 or 5.

N=int(input("Enter the integer N: "))
x=1
s=0
for x in range (N+1 ):
  if (x % 3 == 0) or (x % 5 == 0):
    s=s+x
  x=x+1
print("Total Sum: ",s)

```
