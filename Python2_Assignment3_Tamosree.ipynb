## **Function**

```python
#H1 Write a function to calculate the sum of numbers from 1 to N.

def sum(n):
  s=0
  for i in range(1,n+1):
    s=s+i
  return s

n= int(input("Enter the Number N:"))
print("Total Summation: ",sum(n))
```

```python
#H2 Write a function that returns the square of a number.

def squ(n):
  s=n*n
  return s

n= int(input("Enter the Number N:"))
print("The square of the number: ",squ(n))
```

```python
#H3 Write a function that checks whether a number is even/odd.

def check (n):
  if n % 2==0:
    print ("The number is EVEN")
  else:
    print ("The number is ODD")

n= int(input("Enter the Number :"))
check (n)
```

```python
#H4 Write a function that returns the larger number from two given numbers.

def max (n , m):
  if n >= m:
    return n
  else:
    return m

n = int(input("Enter First Number :"))
m = int(input("Enter Second Number :"))
print ("The larger number is ",max (n, m))
```

## **Lists**

```python
#H5  	Write a Python program to create a list of 10 integers. Then perform the following operations:
  #Append an element to the list.
  #Insert an element at a specific position.
  #Remove an element from the list.
  #Pop an element from a specific position.

#Append:

fruit = ['Mango', 'Orange', 'Coconut']

print('Original List Of Fruits: ',fruit)

fruit.append('Banana')

print('Updated List Of Fruits(By Append): ',fruit)


# Insert A specific position:

Color = ['Red', 'Black', 'Blue']

print('\nOriginal List Of Color: ',Color)

Color.insert(1 ,'Yellow')

print('Updated List Of Color(Insert in specific position): ',Color)


# Remove :

Name = ['Riya', 'Rahim', 'Banti', 'Rimi']

print('\nOriginal List Of Name: ',Name)

Name.remove('Banti')

print('Updated List Of name (Remove element from list): ',Name)


#Pop:

Name = ['Riya', 'Rahim', 'Banti', 'Rimi']

print('\nOriginal List Of Name: ',Name)

print('Updated List Of name (pop from specific position): ',Name.pop(2))
```

```python
#H6 Write a program to find all unique elements in the list.

Li= [ 2, 7, 9 , 2, 12, 15, 7, 9, 16, 2, 7, 9, 22]

un_Li=list(set(Li))

print("The Unique Elements: ",un_Li)
```

```python
#H7 Write a Python program to calculate the mean of a list of numbers.

Li=[5, 8, 8, 10, 16, 7]

sum= 0

for x in range(6):
  sum= sum + Li[x]

print("The mean of the list: ", sum/6)

```

```python
#H8 Classify Numbers into Even and Odd Lists

Li=[5, 8, 8, 10, 16, 7, 3, 19, 15, 22]

even=[]
odd=[]

for x in range(10):
  if Li[x]%2 == 0:
    even.append(Li[x])
  else:
    odd.append(Li[x])

print("The Even Numbers: ", even)
print("The Odd Numbers: ", odd)
```

```python
#H9 Write a program to find the key in a dictionary that has the maximum value.

my_dict = {'a': 10, 'b': 20, 'c': 5}

max_key = max(my_dict, key=my_dict.get)

print(max_key)

```
